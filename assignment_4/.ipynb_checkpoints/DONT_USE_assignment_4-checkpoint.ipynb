```python
import time

# Function to parse SAM file
def parse_sam(filename):
    sam_dict = {}
    with open(filename, 'r') as f:
        for line in f:
            if line.startswith('@'):
                continue  # skip header
            parts = line.strip().split('\t')
            if len(parts) < 5:
                continue
            chrom = parts[2]
            if chrom == '*' or chrom == '':
                continue  # unmapped
            start_pos = int(parts[3])
            if chrom not in sam_dict:
                sam_dict[chrom] = []
            sam_dict[chrom].append(start_pos)
    return sam_dict

# Function to parse GTF file
def parse_gtf(filename):
    exon_list = []
    with open(filename, 'r') as f:
        for line in f:
            if line.startswith('#'):
                continue
            parts = line.strip().split('\t')
            if parts[2] != 'exon':
                continue  # only exons
            chrom = parts[0]
            start = int(parts[3])
            end = int(parts[4])
            # Extract gene_name
            attributes = parts[8]
            gene_name = ''
            for attr in attributes.split(';'):
                if 'gene_name' in attr:
                    gene_name = attr.split('"')[1]
                    break
            if gene_name:
                exon_list.append((chrom, start, end, gene_name))
    return exon_list

# Count reads without sorting
def count_reads_unsorted(sam_dict, exon_list):
    gene_counts = {}
    for chrom, start, end, gene in exon_list:
        if chrom not in sam_dict:
            continue
        count = 0
        for pos in sam_dict[chrom]:
            if start <= pos <= end:
                count += 1
        if gene not in gene_counts:
            gene_counts[gene] = 0
        gene_counts[gene] += count
    return gene_counts

# Count reads with sorted coordinates
def count_reads_sorted(sam_dict, exon_list):
    gene_counts = {}
    sorted_sam = {}
    for chrom in sam_dict:
        sorted_sam[chrom] = sorted(sam_dict[chrom])

    for chrom, start, end, gene in exon_list:
        if chrom not in sorted_sam:
            continue
        count = 0
        pos_list = sorted_sam[chrom]
        i = 0
        while i < len(pos_list):
            if pos_list[i] < start:
                i += 1
            elif start <= pos_list[i] <= end:
                count += 1
                i += 1
            else:  # pos > end
                break
        if gene not in gene_counts:
            gene_counts[gene] = 0
        gene_counts[gene] += count
    return gene_counts

# Main execution
sam_filename = 'sample.sam'
gtf_filename = 'Arabidopsis.gtf'

# Parse files
sam_dict = parse_sam(sam_filename)
exon_list = parse_gtf(gtf_filename)

# Unsorted counting
start_time = time.process_time()
gene_counts_unsorted = count_reads_unsorted(sam_dict, exon_list)
unsorted_time = time.process_time() - start_time

print("Counts without sorting:")
print(gene_counts_unsorted)
print(f"Time without sorting: {unsorted_time:.6f} seconds")

# Sorted counting
start_time = time.process_time()
gene_counts_sorted = count_reads_sorted(sam_dict, exon_list)
sorted_time = time.process_time() - start_time

print("\nCounts with sorting:")
print(gene_counts_sorted)
print(f"Time with sorting: {sorted_time:.6f} seconds")

# Simple explanation
if sorted_time < unsorted_time:
    print("\nSorting improves speed because we can stop checking once we pass the exon end position.")
else:
    print("\nSorting did not help much probably because of small data size, but generally improves performance for large datasets.")
```
